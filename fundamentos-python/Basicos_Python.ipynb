```python
# Esto es un comentario: Python lo ignora. Pero sirve de guía para recordar lo que estábamos haciendo.

# 1. Creación de variables
capital_inicial = 1000
tasa_interes = 0.05
periodos = 3

# 2. Imprimir las variables
print("Capital Inicial:", capital_inicial)
print(f"Tasa de interés: {tasa_interes * 100}%") # Esta expresión permite ajustar el formato del numero a porcentaje.
```

## 2. Operaciones Matemáticas
Python sigue el orden estándar de operaciones (PEMDAS).
Vamos a calcular el **Valor Futuro (FV)** usando la fórmula de interés compuesto:

$$FV = PV \times (1 + r)^n$$

Donde:
* $PV$: Valor Presente
* $r$: Tasa de interés
* $n$: Número de periodos

```python
# Cálculo del Valor Futuro
valor_futuro = capital_inicial * (1 + tasa_interes) ** periodos

# Nota: En Python, la potencia se escribe con doble asterisco (**)

print(f"El valor futuro de la inversión es: ${valor_futuro:.2f}")
# El :.2f sirve para redondear a 2 decimales
```

## 3. Listas (Series de Datos)

En finanzas, rara vez trabajamos con un solo dato. Usualmente tenemos series de precios o portafolios de activos. Las **listas** nos permiten guardar múltiples elementos.

```python
# Una lista de precios de cierre de una acción ficticia
precios_cierre = [100, 102, 101, 105, 107]

# Una lista de tickers
portafolio = ["AAPL", "GOOGL", "TSLA", "AMZN"]

print("Precios:", precios_cierre)
print("Primer activo del portafolio:", portafolio[0]) # Python cuenta desde 0

# Calcular el retorno simple: (Precio Final - Precio Inicial) / Precio Inicial
retorno = (precios_cierre[-1] - precios_cierre[0]) / precios_cierre[0] # Ek indice -1 hace referencia al ultimo elemento de la lista.
print(f"Retorno del periodo: {retorno:.2%}")
```

## 4. Toma de Decisiones (If / Else)
Podemos programar reglas lógicas. Por ejemplo, una regla simple de compra/venta.

```python
precio_actual = 107
precio_objetivo = 110
stop_loss = 95

if precio_actual >= precio_objetivo:
    decision = "VENDER (Tomar ganancias)"
elif precio_actual <= stop_loss:
    decision = "VENDER (Cortar pérdidas)"
else:
    decision = "MANTENER (Hold)"

print(f"La recomendación algorítmica es: {decision}")
```

## 6. Crear una base de datos

En Pandas, una tabla de datos se llama **DataFrame**. Tiene filas y columnas, igual que en **R**.

Vamos a crear una pequeña base de datos de un portafolio manualmente:

```python
# Primero creamos un diccionario con los datos que vamos a emplear en la base de datos

datos_portafolio = {'Ticker': ['AAPL', 'MSFT', 'TSLA', 'AMZN'], 
                    'Empresa': ['Apple', 'Microsoft', 'Tesla', 'Amazon'],
                    'Precio': [150.5, 310.2, 750.0, 3300.0],
                    'Sector': ['Tecnología', 'Tecnología', 'Automotriz', 'E-commerce']}

# Ahora la convertimos en una base de datos

df_acciones = pd.DataFrame(datos_portafolio)

# Y la visualizamos

df_acciones
```

### Operaciones con bases de datos

Se pueden seleccionar columnas especificas o hacer cálculos rápidos.

```python
# Seleccionar solo la columna de precios
print("Precios del portafolio:")
print(df_acciones['Precio'])
```

También se puede usar la posición

```python
print(df_acciones.iloc[1,:]) # Traer la información de la segunda fila
```

```python
print(df_acciones.iloc[:,3]) # Traer la información de la cuarta columna.
```

```python
print(df_acciones.iloc[1,2]) # Información de la fila 2 y columna 3
```

```python
# Calcular estadísticas rápidas

promedio = df_acciones['Precio'].mean()
print(f"Precio promedio de la acción: ${promedio:.2f}")
```

## 7. Cargar Archivos Externos (CSV)

En la vida real, descargamos los datos de diferentes plataformas.

Vamos a ver como usamos **pandas** para cargar una base de datos.

```python
# Vamos a cargar un archivo csv en una nueva base de datos de Python. El archivo se encuentra en un repositorio publico de GitHub, por lo cual no es necesario descargarlo manualmente.

url_file = "https://raw.githubusercontent.com/salas317/Data/main/GRF/Pasajeros.csv"
df = pd.read_csv(filepath_or_buffer = url_file, sep = ";")

df.head()
```

```python
# Arreglemos la fecha

df = pd.read_csv(filepath_or_buffer = url_file, sep = ";", parse_dates = ["mes"], dayfirst = True)

df.head()
```

```python
df.info()
```

```python
# Arreglemos la fecha y que sea el indice

df = pd.read_csv(filepath_or_buffer = url_file, sep = ";", parse_dates = ["mes"], dayfirst = True, index_col = "mes")

df.head()
```
